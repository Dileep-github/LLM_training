Minimum Edit Distance

```python
#define the strings
s1 = "kitten"
s2 = "sitting"

#define the cost for each operation
ins_cost=1
del_cost=1
sub_cost=2

#calculate the length of each string
len_s1 = len(s1)
len_s2 = len(s2)

```

```python
# Initialize the distance matrix
dp = [[0 for _ in range(len_s2 + 1)] for _ in range(len_s1 + 1)]
```

```python
dp
```

```python
# Base cases
for i in range(len_s1 + 1):
    dp[i][0] = i * del_cost
for j in range(len_s2 + 1):
    dp[0][j] = j * ins_cost
```

```python
for i in range(1, len_s1 + 1):
        for j in range(1, len_s2 + 1):
            if s1[i - 1] == s2[j - 1]:
                cost = 0
            else:
                cost = sub_cost

            dp[i][j] = min(
                dp[i - 1][j] + del_cost,     # Deletion
                dp[i][j - 1] + ins_cost,     # Insertion
                dp[i - 1][j - 1] + cost      # Substitution
            )
```

```python
dp
```

```python
distance = dp[len_s1][len_s2]
```

```python
print(f"Custom Levenshtein distance (substitution=2) between '{s1}' and '{s2}' is {distance}")
```

The following function estimates the log-probability of a sentence using an n-gram language model.

1. It breaks the sentence into n-grams (e.g., bigrams, trigrams).

2. For each n-gram, it looks up its probability using the counts of:

    2a. the n-gram itself (ngram_counts)

    2b. its (n−1)-gram context (lower_order_counts)

3. Then it adds the log of each probability to get the total sentence log-probability.


$p(abcd)=p(a|<s>)*p(b|a)*p(c|b)*p(d|c)$,
if we have $count(a)$ and $count(ab)$
Then, $p(b|a) = count(ab)/ count(a)$

```python
# Calculate probability of a sentence
def sentence_log_probability(sentence, ngram_counts, lower_order_counts, n):
    tokens = nltk.word_tokenize(sentence.lower())#preprocessing
    n_grams = list(ngrams(tokens, n))#from query sentence
    log_prob = 0.0
    for gram in n_grams:
        count = ngram_counts.get(gram, 0)
        denom = lower_order_counts.get(gram[:-1], 0) if n > 1 else sum(ngram_counts.values())
        if count == 0 or denom == 0:
            return float('-inf')  # log(0) is undefined
        prob = count / denom
        log_prob += math.log(prob)
    return log_prob
```

In natural language processing, Laplace smoothing (also called add-one smoothing) is a technique used to handle zero probabilities in language models.

For example, if a bigram like "dog runs" doesn't appear in your training data, the model would assign it a probability of 0 — which can break things like perplexity or log-probability calculations. Laplace smoothing solves this by adding 1 to every count, ensuring that no probability is ever zero.

```python
def laplace_smoothing(ngram_counts, lower_order_counts, n, vocab):
    smoothed_probs = defaultdict(float)
    V = len(vocab)

    # Generate all possible n-grams from vocab
    if n == 1:
        all_ngrams = [(w,) for w in vocab]
    else:
        all_contexts = list(product(vocab, repeat=n-1))
        all_ngrams = [context + (w,) for context in all_contexts for w in vocab]

    for ngram in all_ngrams:
        context = ngram[:-1] if n > 1 else ()
        count_ngram = ngram_counts.get(ngram, 0)
        count_context = lower_order_counts.get(context, 0)
        prob = (count_ngram + 1) / (count_context + V)
        smoothed_probs[ngram] = prob

    return smoothed_probs
```

```python
smoothed_bigram_probs = laplace_smoothing(bigram_counts, unigram_counts, 2, vocab)
smoothed_trigram_probs = laplace_smoothing(trigram_counts, bigram_counts, 3, vocab)
```

```python
# Calculate Perplexity
def perplexity(test_tokens, ngram_counts, lower_order_counts, n):
    n_grams = list(ngrams(test_tokens, n))
    N = len(n_grams)
    log_prob = 0
    for gram in n_grams:
        count = ngram_counts.get(gram, 0)
        denom = lower_order_counts.get(gram[:-1], 0) if n > 1 else sum(ngram_counts.values())
        if count == 0 or denom == 0:
            return float('inf')
        prob = count / denom
        log_prob += math.log(prob)
    return math.exp(-log_prob / N)
```

```python
def smoothed_perplexity(test_tokens, smoothed_probs, n):
    n_grams = list(ngrams(test_tokens, n))
    N = len(n_grams)
    log_prob = 0
    for gram in n_grams:
        prob = smoothed_probs.get(gram, 1 / V)
        log_prob += math.log(prob)
    return math.exp(-log_prob / N)
```

```python
test_sentence = "the cat sat on the"
test_sentence1 = "At night, the wind whispered secrets through the trees"
test_tokens = nltk.word_tokenize(test_sentence1.lower())
```

```python
print("Perplexity without smoothing (bigrams):", perplexity(test_tokens, bigram_counts, unigram_counts, 2))
print("Perplexity with smoothing (bigrams):", smoothed_perplexity(test_tokens, smoothed_bigram_probs, 2))
```

```python
# Predict next word
def predict_next_word(prefix, ngram_probs, n):
    prefix_tokens = tuple(nltk.word_tokenize(prefix.lower())[-(n-1):])
    candidates = {k[-1]: v for k, v in ngram_probs.items() if k[:-1] == prefix_tokens}
    sorted_candidates = sorted(candidates.items(), key=lambda x: x[1], reverse=True)
    return sorted_candidates[:5]  # Top 5 predictions
```

```python
print("Next word predictions (bigrams):", predict_next_word("At night, the wind whispered", smoothed_bigram_probs, 2))
print("Next word predictions (trigrams):", predict_next_word("At night, the ", smoothed_trigram_probs, 3))
```

```python
# Probability of a sentence with smoothing
def smoothed_sentence_probability(sentence, smoothed_probs, n):
    tokens = nltk.word_tokenize(sentence.lower())
    n_grams = list(ngrams(tokens, n))
    prob = 1.0
    for gram in n_grams:
        prob *= smoothed_probs.get(gram, 1 / V)
    return prob

print("Smoothed sentence probability:", smoothed_sentence_probability(test_sentence, smoothed_bigram_probs, 2))
```
