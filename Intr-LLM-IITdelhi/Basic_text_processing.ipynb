**Stemming**


---

**Definition:** Stemming is the process of reducing words to their root or base form by removing prefixes or suffixes, **often without considering the word's grammatical context**. It helps in grouping similar words for text analysis, but the resulting stemmed words may not always be valid dictionary words.

**Example:** Using the Porter Stemmer in Python:

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ["running", "runner", "runs", "easily", "better"]
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

**spaCy:** Unlike NLTK, spaCy does not support stemming because it focuses on production-level accuracy and prefers lemmatization, which uses grammatical context and dictionaries for more precise results.

**Lemmatization**


---




*   **Definition:** Lemmatization reduces words to their base or dictionary form (lemma) while considering the word's context (e.g., part of speech). Unlike stemming, lemmatization produces valid words. For example, "running" becomes "run," and "better" becomes "good" (if contextually appropriate).
*   **Context-Aware:** Lemmatization uses linguistic rules and dictionaries, making it more accurate than stemming. For instance, "rocks" becomes "rock," and "better" becomes "good" (adjective form).



**Lemmatization** using NLTK

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

# Initialize lemmatizer
lemmatizer = WordNetLemmatizer()

# Words to lemmatize
words = ["running", "runner", "best", "rocks", "went"]

# Lemmatize with part-of-speech (POS) tagging
lemmatized_words = [lemmatizer.lemmatize(word, pos="v") for word in words]  # 'v' for verbs
print("Lemmatized Words (NLTK):", lemmatized_words)
# Output: ['run', 'best', 'rock', 'go']
```

Using **spaCy** for Lemmatization

```python
import spacy

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Text to process
text = "The rocks are running better than before."

# Process text
doc = nlp(text)

print(type(doc))
# Extract lemmas
for token in doc:
  print(token.text, token.lemma_)
lemmatized_words = [(token.text, token.lemma_) for token in doc]
print("Lemmatized Words (spaCy):", lemmatized_words)

words = ["running", "runner", "best", "rocks", "went"]
# Process each word individually with spaCy to create Token objects
lemmatized_words = [(word, nlp(word)[0].lemma_) for word in words]
print("Lemmatized Words (spaCy):", lemmatized_words)
```

**What is Stop Word Removal?**

*   **Definition:** Stop word removal is a preprocessing step in Natural Language Processing (NLP) where commonly occurring words in a language (e.g., "the," "is," "and") are filtered out from text data. These words typically carry little semantic value and are removed to focus on more meaningful terms in the text.
*   **Purpose:**

1.   **Reduce Noise:** Stop words can clutter the data without adding significant meaning, so removing them helps reduce noise in textual analysis.
2.   **Improve Efficiency:** Eliminating stop words reduces the size of the text data, speeding up processing and reducing computational resources required for NLP tasks.
3.  **Enhance Model Accuracy:** By focusing on semantically important words, stop word removal can improve the performance of models like text classification, topic modeling, and search engines.





**Stopword Removal with NLTK**

```python
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
text = "I am just looooving NLP!!! with @john #NLPdays"
# Word Tokenization
word_tokens = word_tokenize(text)
filtered_words = [word for word in word_tokens if word.lower() not in stop_words]
print("Filtered Words:", filtered_words)
```

**Stopword Removal with spaCy**

```python
# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

# Text to process
text = "I am just looooving NLP!!! with @john #NLPdays"

# Process text
doc = nlp(text)
filtered_words = [token.text for token in doc if not token.is_stop]
print("Filtered Words:", filtered_words)
```

**NER using NLTK**

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
from nltk import word_tokenize, pos_tag, ne_chunk

# Download necessary resources
nltk.download('punkt')
nltk.download('maxent_ne_chunker')
nltk.download('words')

# Sample sentence
sentence = "Apple is looking at buying U.K. startup for $1 billion."

# Tokenize and POS tag the sentence
tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)

# Perform Named Entity Recognition
named_entities = ne_chunk(pos_tags)
print(named_entities)
```

**POS tagging using NLTK**

```python
import nltk

# Download necessary resources
nltk.download('averaged_perceptron_tagger')

# Sample sentence
sentence = "The quick brown fox jumps over the lazy dog."

# Tokenize the sentence
tokens = nltk.word_tokenize(sentence)

# Perform POS tagging
pos_tags = nltk.pos_tag(tokens)
print(pos_tags)
```

```python
# Named Entity Recognition (NER) with spaCy:

text = "Barack Obama was the 44th President of the United States."
doc = nlp(text)
print("NER")
print([(ent.text, ent.label_) for ent in doc.ents])

# Part-of-Speech (POS) Tagging:
print("POS")
for token in doc:
    print(token.text, token.pos_)

```
