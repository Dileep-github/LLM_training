```python
# --- Concept: What is WordNet? ---
# WordNet is a large lexical database of English.
# Nouns, verbs, adjectives, and adverbs are grouped into sets of cognitive synonyms (called synsets),
# each expressing a distinct concept. Synsets are interlinked by means of conceptual-semantic and lexical relations.

# --- Concept: What is a Synset? ---
# A 'synset' (short for "synonym set") is a group of synonymous words that share a common meaning.
# Each synset includes:
# - A unique identifier (e.g., 'dog.n.01')
# - A definition (gloss)
# - Example sentences (optional)
# - Relationships to other synsets (e.g., hypernyms, hyponyms)

print("\n--- Exploring Synsets ---")
word = 'bank'
synsets = wn.synsets(word)
print(f"All synsets for the word '{word}':\n")
for i, syn in enumerate(synsets):
    print(f"{i + 1}. {syn.name()} - {syn.definition()}")

```

```python
# --- Word Sense Similarity ---
# WordNet allows measuring semantic similarity between two synsets.
# Common similarity metrics include:
# - Path similarity: based on shortest path in the hierarchy
# - Wu-Palmer similarity: based on depth of synsets in the taxonomy

print("\n--- Word Sense Similarity ---")
# Let's compare two related senses: 'dog.n.01' and 'wolf.n.01'
syn1 = wn.synset('dog.n.01')
syn2 = wn.synset('wolf.n.01')

print(f"Synset 1: {syn1.name()} - {syn1.definition()}")
print(f"Synset 2: {syn2.name()} - {syn2.definition()}")
print("\nPath Similarity:", syn1.path_similarity(syn2))
print("Wu-Palmer Similarity:", syn1.wup_similarity(syn2))

```

```python
# Another example: 'car.n.01' and 'bicycle.n.01'
syn3 = wn.synset('car.n.01')
syn4 = wn.synset('bicycle.n.01')
print("\nPath Similarity (car vs. bicycle):", syn3.path_similarity(syn4))
print("Wu-Palmer Similarity (car vs. bicycle):", syn3.wup_similarity(syn4))

```

```python
# --- Closest Synset Matching ---
# A function to get the maximum similarity between all synset pairs of two words

def max_similarity(word1, word2, metric='path'):
    synsets1 = wn.synsets(word1)
    synsets2 = wn.synsets(word2)
    max_sim = 0.0
    for s1 in synsets1:
        for s2 in synsets2:
            if metric == 'path':
                sim = wn.path_similarity(s1, s2)
            elif metric == 'wup':
                sim = wn.wup_similarity(s1, s2)
            else:
                raise ValueError("Unsupported similarity metric")
            if sim is not None and sim > max_sim:
                max_sim = sim
    return max_sim

print("\n--- Max Similarity Between Words ---")
print("Word: 'car' vs 'automobile':", max_similarity('car', 'automobile', metric='wup'))
print("Word: 'car' vs 'bicycle':", max_similarity('car', 'bicycle', metric='wup'))
print("Word: 'dog' vs 'cat':", max_similarity('dog', 'cat', metric='wup'))
```

```python
corpus = [
    "NLP is fun and exciting",
    "NLP and machine learning are the future",
    "machine learning is powerful"
]
```

```python
import math
from collections import Counter

# Preprocess and tokenize
def preprocess(doc):
    return doc.lower().split()

tokenized_corpus = [preprocess(doc) for doc in corpus]
```

```python
print(tokenized_corpus)
```

```python
# Compute TF for each document
def compute_tf(tokenized_doc):
    tf_dict = {}
    word_counts = Counter(tokenized_doc)
    total_words = len(tokenized_doc)
    for word, count in word_counts.items():
        tf_dict[word] = count / total_words
    return tf_dict

tf_corpus = [compute_tf(doc) for doc in tokenized_corpus]

print("TF for each document:")
for i, tf in enumerate(tf_corpus):
    print(f"Doc {i+1} TF:", tf)
```

```python
# Build the full vocabulary
vocab = set(word for doc in tokenized_corpus for word in doc)

# Compute IDF
def compute_idf(tokenized_corpus):
    N = len(tokenized_corpus)
    idf_dict = {}
    for word in vocab:
        doc_freq = sum(1 for doc in tokenized_corpus if word in doc)
        idf_dict[word] = math.log(N / (1 + doc_freq)) + 1  # Smoothed IDF
    return idf_dict

idf = compute_idf(tokenized_corpus)

print("\nIDF for each word:")
for word, val in idf.items():
    print(f"{word}: {val:.4f}")
```

```python
# Compute TF-IDF
def compute_tfidf(tf_doc, idf):
    tfidf = {}
    for word, tf_val in tf_doc.items():
        tfidf[word] = tf_val * idf.get(word, 0.0)
    return tfidf

tfidf_corpus = [compute_tfidf(tf, idf) for tf in tf_corpus]

print("\nTF-IDF for each document:")
for i, tfidf in enumerate(tfidf_corpus):
    print(f"Doc {i+1} TF-IDF:", tfidf)
```

```python
# Sort vocabulary to maintain order
sorted_vocab = sorted(vocab)

def vectorize_tfidf(tfidf, sorted_vocab):
    return [tfidf.get(word, 0.0) for word in sorted_vocab]

doc_vectors = [vectorize_tfidf(tfidf, sorted_vocab) for tfidf in tfidf_corpus]

print("\nTF-IDF Vectors:")
print("Vocabulary:", sorted_vocab)
for i, vec in enumerate(doc_vectors):
    print(f"Doc {i+1} vector:", vec)
```

```python
######Computing PMI(Point wise Mutual Information) from a Document Corpus
from collections import defaultdict

window_size = 2
word_counts = Counter()
cooccurrence_counts = defaultdict(int)

# Count words and co-occurring word pairs
for doc in tokenized_corpus:
    for i, word in enumerate(doc):
        word_counts[word] += 1
        window_start = max(0, i - window_size)
        window_end = min(len(doc), i + window_size + 1)
        for j in range(window_start, window_end):
            if i != j:
                pair = tuple(sorted([word, doc[j]]))  # sort to avoid (a,b) vs (b,a)
                cooccurrence_counts[pair] += 1
```

```python
# Total number of words and co-occurrence pairs
total_word_count = sum(word_counts.values())
total_cooccurrences = sum(cooccurrence_counts.values())

# Compute PMI for each co-occurring pair
pmi_scores = {}
for pair, cooc_count in cooccurrence_counts.items():
    w1, w2 = pair
    p_w1 = word_counts[w1] / total_word_count
    p_w2 = word_counts[w2] / total_word_count
    p_w1_w2 = cooc_count / total_cooccurrences
    pmi = math.log2(p_w1_w2 / (p_w1 * p_w2))
    pmi_scores[pair] = pmi
```

```python
print("PMI Scores (Top Associations):")
for pair, score in sorted(pmi_scores.items(), key=lambda x: -x[1])[:10]:
    print(f"{pair}: {score:.4f}")
```

```python
#############One-hot Encoding###########
```

```python
corpus = [
    "NLP is fun and exciting",
    "NLP and machine learning are the future",
    "machine learning is powerful"
]
```

```python
# Tokenize and lowercase
tokenized_corpus = [doc.lower().split() for doc in corpus]

# Flatten and build vocabulary
from itertools import chain
all_tokens = list(chain(*tokenized_corpus))
vocab = sorted(set(all_tokens))
vocab_size = len(vocab)

# Create word-to-index and index-to-word mappings
word2idx = {word: idx for idx, word in enumerate(vocab)}
idx2word = {idx: word for word, idx in word2idx.items()}
```

```python
import numpy as np

def one_hot_vector(word, word2idx, vocab_size):
    vec = np.zeros(vocab_size)
    idx = word2idx[word]
    vec[idx] = 1
    return vec
```

```python
# Display one-hot vectors for a few words
for word in ["nlp", "machine", "learning", "fun"]:
    vec = one_hot_vector(word, word2idx, vocab_size)
    print(f"{word}: {vec.astype(int)}")
```
